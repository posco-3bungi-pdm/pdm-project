`isna()` 는 빈칸이면 True를 돌려줍니다. 여기에 `.sum()` 을 붙이면 개수가 나옵니다.

```python
# TODO: 컬럼별 빈칸 개수를 출력하세요.
# 힌트: df.isna().sum()
empty = df.isna().sum()
print(empty)

print("===============")

# TODO: 비율(%)로도 확인해 보세요.
# 힌트: (df.isna().mean() * 100).round(2)
empty_per = (df.isna().mean() * 100).round(2)
print(empty_per)
```

`nunique()` 는 그 컬럼에 **서로 다른 값이 몇 개 있는지** 세어줍니다.
이 값이 1이면 처음부터 끝까지 같은 값만 들어 있다는 뜻입니다.

```python
# TODO: 컬럼별 고유값 개수를 확인하세요.
# 힌트: df.nunique()
count = df.nunique()
print(count)

print("===============")

# TODO: 고유값이 1개인 컬럼만 골라내세요.
# 힌트: const_cols = df.columns[df.nunique() <= 1]
#       print(list(const_cols))
const_cols = df.columns[df.nunique() <= 1]
print(list(const_cols))
```
