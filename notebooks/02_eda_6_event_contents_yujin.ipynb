# 조업이벤트.csv 파일 분석

#### 담당: 황유진 . 관련이슈: # 29

```python
import pandas as pd

df = pd.read_csv("../data/G-02_조업이벤트.csv", encoding="utf-8-sig")
df.head()
```

## 행 수, 컬럼, 결측, 중복 확인

```python
print("행 수:, 컬럼 수:", df.shape)

print("컬럼명:", df.columns)

print("결측값:")
print(df.isna().sum())

print("중복행:", df.duplicated().sum())
```

## EVT_TYPE 종류별 건수

```python
print(df["EVT_TYPE"].value_counts())
```

## PLANT_CD 종류별 건수

```python
print(df["PLANT_CD"].value_counts())
```

## PLANT_CD와 EVT_TYPE 같이 보기

```python
pd.crosstab(df["PLANT_CD"], df["EVT_TYPE"])
```

## 강종변경

```python
grade = df[df["EVT_TYPE"] == "강종변경"]

print("변경 전:")
print(grade["BEF_VAL"].unique())

print("변경 후:")
print(grade["AFT_VAL"].unique())
```

## 원료배합변경

```python
material = df[df["EVT_TYPE"] == "원료배합변경"]

print("변경 전:")
print(material["BEF_VAL"].unique())

print("변경 후:")
print(material["AFT_VAL"].unique())
```

## 조업률변경

```python
rate = df[df["EVT_TYPE"] == "조업률변경"]

print("변경 전:")
print(rate["BEF_VAL"].unique()[:10])

print("변경 후:")
print(rate["AFT_VAL"].unique()[:10])
```

## 월별 이벤트 건수

```python
df["EVT_DT"] = pd.to_datetime(df["EVT_DT"])
```

```python
month_count = df["EVT_DT"].dt.month.value_counts().sort_index()
print(month_count)
```

### 주요 확인 내용

- 강종변경이 134건으로 가장 많았다.
- 조업률변경은 109건, 원료배합변경은 20건이었다.
- CR1에서는 강종변경 이벤트만 발생했다.
- BF1, BF2에서는 원료배합변경과 조업률변경이 발생했다.
- SP1, SP2에서는 조업률변경만 발생했다.
- BEF_VAL과 AFT_VAL은 EVT_TYPE에 따라 의미가 달라진다.
- 월별 이벤트는 특정 한 달에만 몰리지 않고 연중 분포되어 있다.
